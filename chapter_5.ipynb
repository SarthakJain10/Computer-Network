# Transport Layer

## Introduction
The **Transport Layer** is the **4th layer** in the OSI model and the **2nd layer** in the TCP/IP model. It provides **end-to-end communication** between applications running on different hosts. This layer ensures that data is delivered **reliably, in order, and without errors** from the source to the destination.

---

## Responsibilities of the Transport Layer

### Key Roles:
- **End-to-End Communication**: Establishes a logical connection between two applications.
- **Reliability**: Ensures data is delivered correctly using mechanisms like **acknowledgments** and **retransmissions**.
- **Flow Control**: Adjusts the data transmission rate to prevent overwhelming the receiver.
- **Congestion Control**: Manages network traffic to avoid congestion and data loss.
- **Multiplexing/Demultiplexing**: Allows multiple applications to share a single network connection.
- **Error Control**: Detects and corrects errors in transmitted data.

---

## Port Numbers

### Definition
A **port number** is a **16-bit unsigned integer** (0–65535) used to identify specific processes or services running on a host.

### Purpose
- Enables **multiple applications** to use the network simultaneously.
- Helps the **Transport Layer** direct incoming data to the correct application.

### Examples
- **HTTP**: Port 80
- **HTTPS**: Port 443
- **FTP**: Port 21
- **SSH**: Port 22

---

## IP Address vs Port Number

| Feature          | IP Address                          | Port Number                     |
|------------------|-------------------------------------|---------------------------------|
| **Purpose**      | Identifies a **host** on a network  | Identifies a **process** on a host |
| **Format**       | 32-bit (IPv4) or 128-bit (IPv6)      | 16-bit (0–65535)                |
| **Example**      | `192.168.1.1`                      | `80` (HTTP)                     |
| **Layer**        | **Network Layer**                   | **Transport Layer**             |

---

## IANA Port Number Ranges

| Range             | Port Numbers       | Usage                          |
|-------------------|--------------------|--------------------------------|
| **Well-known**    | 0–1023             | Reserved for system processes (e.g., HTTP, FTP) |
| **Registered**    | 1024–49151         | Assigned to user applications (e.g., MySQL, PostgreSQL) |
| **Dynamic/Private** | 49152–65535      | Temporary use by client applications |

---

## Socket Address

### Definition
A **socket address** is the combination of an **IP address** and a **port number** (e.g., `192.168.1.1:80`). It uniquely identifies a **specific process** on a **specific host**.

### Role in Communication
- Enables precise delivery of data to the correct application on the correct host.
- Used by protocols like **TCP** and **UDP** to establish connections.

---

## Multiplexing and Demultiplexing

### Concept
- **Multiplexing**: Combining data from multiple applications into a single stream for transmission.
- **Demultiplexing**: Separating the combined data stream back into individual application data at the receiver.

### Sender-Side (Multiplexing)
- Gathers data from multiple sockets.
- Adds **header information** (e.g., port numbers) to identify the source/destination.
- Sends the combined data to the **Network Layer**.

### Receiver-Side (Demultiplexing)
- Receives the combined data from the **Network Layer**.
- Uses **header information** (e.g., port numbers) to direct data to the correct application.

---

## Summary Points
- The **Transport Layer** ensures **reliable, end-to-end communication**.
- **Port numbers** identify applications, while **IP addresses** identify hosts.
- **Socket addresses** (IP + Port) enable precise data delivery.
- **Multiplexing/Demultiplexing** allows multiple applications to share a network connection efficiently.


# User Datagram Protocol (UDP)

## Introduction
The **User Datagram Protocol (UDP)** is a **connectionless**, **unreliable** transport layer protocol. It is used for sending **datagrams** (messages) between applications over an IP network. UDP is defined in **RFC 768** and is an alternative to **TCP**, offering lower overhead and faster transmission.

---

## Features of UDP

### Key Characteristics:
- **Connectionless**: No handshake or connection setup is required before sending data.
- **Unreliable**: No acknowledgments, retransmissions, or guarantees of delivery.
- **Low Overhead**: Minimal header size (only 8 bytes).
- **Fast**: Suitable for applications where speed is more important than reliability.
- **No Flow Control**: Sender does not adjust transmission rate based on receiver capacity.
- **No Congestion Control**: Does not monitor or manage network congestion.

---

## UDP Datagram Format

The UDP datagram consists of a **header** and **data**. The header is **8 bytes** long and contains the following fields:

| Field               | Size (bits) | Description                                                                 |
|---------------------|-------------|-----------------------------------------------------------------------------|
| **Source Port**     | 16          | Identifies the sender’s application process.                               |
| **Destination Port**| 16          | Identifies the receiver’s application process.                             |
| **Length**          | 16          | Total length of the UDP datagram (header + data), in bytes.                |
| **Checksum**        | 16          | Used for error detection in the header and data.                           |
| **Data**            | Variable    | The actual application data being transmitted.                             |

---

## Checksum in UDP

### Purpose
- Detects **errors** in the transmitted datagram.
- Ensures data integrity between sender and receiver.

### Working
- The sender calculates the checksum by treating the datagram (header + data) as a sequence of **16-bit words**.
- The checksum is the **16-bit one’s complement** of the sum of all words.
- The receiver recalculates the checksum and compares it with the received checksum. If they don’t match, the datagram is discarded.

---

## UDP Operation

### How Communication Happens
1. **No Connection Setup**: UDP does not establish a connection before sending data.
2. **Datagram Transmission**: The sender encapsulates the application data in a UDP datagram and sends it to the destination.
3. **No Acknowledgment**: The sender does not wait for acknowledgments from the receiver.
4. **No Retransmission**: Lost or corrupted datagrams are **not retransmitted**.

### Example
- A **DNS query** is sent using UDP. The client sends a request, and the server responds without any connection setup or acknowledgment.

---

## Well-Known Ports in UDP

| Application       | Port Number |
|-------------------|-------------|
| **DNS**           | 53          |
| **DHCP**          | 67, 68      |
| **TFTP**          | 69          |
| **SNMP**          | 161         |
| **RIP**           | 520         |
| **VoIP (RTP)**    | 5004, 5005  |

---

## Applications of UDP

### Where UDP is Used
- **Real-Time Applications**: Video conferencing, live streaming, online gaming (e.g., **VoIP, RTP**).
- **DNS (Domain Name System)**: Fast resolution of domain names to IP addresses.
- **DHCP (Dynamic Host Configuration Protocol)**: Assigning IP addresses dynamically.
- **TFTP (Trivial File Transfer Protocol)**: Simple file transfers.
- **SNMP (Simple Network Management Protocol)**: Network management and monitoring.

---

## Advantages and Disadvantages

### Advantages
- **Low Overhead**: Minimal header size (8 bytes).
- **Fast Transmission**: No connection setup or acknowledgments.
- **Efficient for Small Messages**: Ideal for query-response protocols (e.g., DNS).

### Disadvantages
- **Unreliable**: No guarantee of delivery, ordering, or error correction.
- **No Flow Control**: Can overwhelm the receiver with data.
- **No Congestion Control**: Does not adapt to network congestion.

---

## Summary Points
- UDP is a **connectionless**, **unreliable** protocol with **low overhead**.
- It uses a **simple datagram format** with minimal fields.
- The **checksum** ensures basic error detection.
- UDP is ideal for **real-time applications** where speed is critical.
- **Well-known ports** like 53 (DNS) and 67/68 (DHCP) are commonly used with UDP.

# Transmission Control Protocol (TCP)

## Introduction
The **Transmission Control Protocol (TCP)** is a **connection-oriented**, **reliable** transport layer protocol. It ensures that data is delivered **error-free, in order, and without loss** between applications over an IP network. TCP is defined in **RFC 793** and is widely used for applications requiring reliable communication, such as web browsing, email, and file transfer.

---

## Features of TCP

### Key Services:
- **Connection-Oriented**: Establishes a connection before data transfer.
- **Reliable**: Guarantees delivery of data using **acknowledgments** and **retransmissions**.
- **Ordered Data Transfer**: Ensures data is delivered in the correct sequence.
- **Flow Control**: Adjusts the data transmission rate to match the receiver’s capacity.
- **Congestion Control**: Manages network traffic to prevent congestion.
- **Full-Duplex Communication**: Allows data to flow in both directions simultaneously.

---

## TCP Segment Format

The TCP segment consists of a **header** and **data**. The header is **20–60 bytes** long and contains the following fields:

| Field                  | Size (bits) | Description                                                                                     |
|------------------------|-------------|-------------------------------------------------------------------------------------------------|
| **Source Port**        | 16          | Identifies the sender’s application process.                                                   |
| **Destination Port**   | 16          | Identifies the receiver’s application process.                                                  |
| **Sequence Number**    | 32          | Identifies the position of the first data byte in the segment.                                  |
| **Acknowledgment Number** | 32      | Indicates the next expected byte from the sender.                                               |
| **Data Offset**        | 4           | Specifies the size of the TCP header.                                                           |
| **Reserved**           | 6           | Reserved for future use.                                                                        |
| **Flags**              | 6           | Control bits (e.g., **SYN, ACK, FIN, RST, PSH, URG**).                                         |
| **Window Size**        | 16          | Indicates the receiver’s available buffer space.                                               |
| **Checksum**           | 16          | Ensures data integrity.                                                                         |
| **Urgent Pointer**     | 16          | Points to urgent data in the segment.                                                          |
| **Options**            | Variable    | Optional fields for additional functionalities (e.g., **Maximum Segment Size**).               |
| **Data**               | Variable    | The actual application data being transmitted.                                                 |

---

## TCP Connection Establishment

### Three-Way Handshake
TCP establishes a connection using a **three-way handshake**:

1. **SYN (Synchronize)**
   - The **client** sends a segment with **SYN=1** and a random **sequence number (Seq=x)**.
   - ```
     Client ---------------------SYN, Seq=x---------------------> Server
     ```

2. **SYN-ACK (Synchronize-Acknowledge)**
   - The **server** responds with **SYN=1**, **ACK=1**, its own **sequence number (Seq=y)**, and an **acknowledgment number (Ack=x+1)**.
   - ```
     Client <-------------------SYN+ACK, Seq=y, Ack=x+1------------ Server
     ```

3. **ACK (Acknowledge)**
   - The **client** sends an **ACK=1** with **Seq=x+1** and **Ack=y+1**.
   - The connection is now established.
   - ```
     Client ---------------------ACK, Seq=x+1, Ack=y+1------------> Server
     ```

---

## TCP Connection Termination

### Four-Way Handshake
TCP terminates a connection using a **four-way handshake**:

1. **FIN (Finish)**
   - The **client** sends a **FIN** segment to the server.
   - ```
     Client ---------------------FIN---------------------> Server
     ```

2. **ACK (Acknowledge)**
   - The **server** acknowledges the FIN with an **ACK**.
   - ```
     Client <-------------------ACK------------------------ Server
     ```

3. **FIN (Finish)**
   - The **server** sends its own **FIN** segment.
   - ```
     Client <-------------------FIN------------------------ Server
     ```

4. **ACK (Acknowledge)**
   - The **client** acknowledges the server’s FIN with an **ACK**.
   - The connection is now closed.
   - ```
     Client ---------------------ACK---------------------> Server
     ```

---

## Flow Control

### Sliding Window Mechanism
- **Flow control** ensures the sender does not overwhelm the receiver.
- The **window size** field in the TCP header indicates the receiver’s available buffer space.
- The sender adjusts its transmission rate based on the **window size** advertised by the receiver.

---

## Error Control

### Acknowledgment and Retransmission
- **Acknowledgment (ACK)**: The receiver sends an **ACK** for successfully received segments.
- **Retransmission**: If an **ACK** is not received within a timeout period, the sender retransmits the segment.
- **Selective Acknowledgment (SACK)**: Allows the receiver to acknowledge specific segments, improving efficiency.

---

## Summary Points
- TCP is a **connection-oriented**, **reliable** protocol.
- Uses a **three-way handshake** for connection establishment and a **four-way handshake** for termination.
- **Flow control** and **error control** ensure efficient and reliable data transfer.
- The **sliding window mechanism** manages the data transmission rate.
- TCP is ideal for applications requiring **reliable, ordered data delivery**, such as **HTTP, FTP, and email**.

# Stream Control Transmission Protocol (SCTP)

## Introduction
The **Stream Control Transmission Protocol (SCTP)** is a **transport layer protocol** designed to combine the best features of **TCP** and **UDP**. It provides **reliable, ordered delivery** of data with **congestion control**, similar to TCP, but also supports **multi-homing** and **multi-streaming**, making it ideal for modern applications like **VoIP, telephony, and real-time data transfer**. SCTP is defined in **RFC 4960**.

---

## Features of SCTP

### Key Services:
- **Reliable Data Transfer**: Ensures data is delivered **error-free** and **in order**.
- **Multi-Homing**: Supports multiple IP addresses for a single endpoint, improving **fault tolerance**.
- **Multi-Streaming**: Allows multiple independent streams within a single association, reducing **head-of-line blocking**.
- **Congestion Control**: Manages network traffic to prevent congestion.
- **Message-Oriented**: Preserves message boundaries, unlike TCP’s byte-stream approach.
- **Partial Reliability**: Allows applications to specify the importance of data, enabling controlled data loss for non-critical messages.

---

## SCTP Packet Format

SCTP uses a **chunk-based structure**, where each packet consists of a **common header** followed by one or more **chunks**. Each chunk contains either **control information** or **user data**.

### Common Header:
| Field               | Size (bits) | Description                                                                 |
|---------------------|-------------|-----------------------------------------------------------------------------|
| **Source Port**     | 16          | Identifies the sender’s port.                                               |
| **Destination Port**| 16          | Identifies the receiver’s port.                                             |
| **Verification Tag**| 32          | Validates the packet belongs to the current association.                    |
| **Checksum**        | 32          | Ensures data integrity.                                                     |

### Chunk Types:
- **DATA**: Carries user data.
- **INIT**: Establishes an association.
- **SACK**: Selective acknowledgment for error control.
- **HEARTBEAT**: Monitors the liveness of the peer.
- **ABORT**: Terminates an association abruptly.

---

## SCTP Association

### Concept
- An **SCTP association** is similar to a **TCP connection** but supports **multi-homing** and **multi-streaming**.
- An association is established using a **four-way handshake** (similar to TCP but more robust).
- Each association can have **multiple streams**, allowing independent data transmission.

---

## Flow Control and Error Control

### Flow Control
- SCTP uses a **sliding window mechanism** similar to TCP.
- The **receiver advertises its buffer space** to the sender, controlling the data transmission rate.

### Error Control
- **Selective Acknowledgment (SACK)**: The receiver acknowledges only the successfully received chunks, enabling efficient retransmission of lost data.
- **Checksum**: Ensures data integrity.
- **Partial Reliability Extension (PR-SCTP)**: Allows applications to specify the reliability level, enabling controlled data loss for non-critical messages.

---

## Summary Points
- SCTP combines the **reliability of TCP** with the **flexibility of UDP**.
- Supports **multi-homing** and **multi-streaming**, reducing head-of-line blocking.
- Uses a **chunk-based packet format** for efficient data transfer.
- Provides **flow control** and **error control** mechanisms similar to TCP but with enhanced features.
- Ideal for **real-time applications** like VoIP and telephony, where reliability and multi-streaming are critical.

# Congestion Control

---

## Introduction

### Definition
**Congestion control** refers to the mechanisms and techniques used to **prevent or manage network congestion**, ensuring efficient and fair use of network resources. Congestion occurs when the **demand for network resources exceeds the available capacity**, leading to degraded performance.

### Causes of Congestion
- **High Traffic Load**: Too many packets being transmitted simultaneously.
- **Insufficient Bandwidth**: Limited capacity of network links.
- **Buffer Overflow**: Routers and switches running out of buffer space.
- **Poor Traffic Management**: Lack of effective routing or scheduling policies.

---

## Network Performance

### Key Metrics
- **Throughput**: The **rate of successful data delivery** over a network, measured in **bits per second (bps)**.
- **Delay**: The **time taken** for a packet to travel from source to destination, including **processing, queuing, transmission, and propagation delays**.
- **Packet Loss**: The **percentage of packets** that fail to reach their destination due to congestion or errors.

---

## Open Loop vs Closed Loop Congestion Control

### Open Loop (Preventive)
- **Definition**: Mechanisms designed to **prevent congestion** before it occurs.
- **Methods**:
  - **Resource Reservation**: Allocating bandwidth in advance.
  - **Admission Control**: Limiting the number of active connections.
  - **Traffic Shaping**: Regulating the rate of data transmission.

### Closed Loop (Reactive)
- **Definition**: Mechanisms that **respond to congestion** after it is detected.
- **Methods**:
  - **Backpressure**: Notifying upstream nodes to slow down transmission.
  - **Choke Packets**: Sending explicit signals to reduce transmission rates.
  - **Adaptive Algorithms**: Dynamically adjusting transmission rates based on network feedback.

---

## Leaky Bucket Algorithm

### Concept
The **Leaky Bucket Algorithm** is a **traffic shaping** mechanism used to control the rate of data transmission and smooth out bursty traffic.

### Working Mechanism
1. **Bucket with a Hole**: Imagine a bucket with a small hole at the bottom. Water (packets) enters the bucket at variable rates but leaks out at a **constant rate**.
2. **Token Generation**: Tokens are generated at a **fixed rate** and added to the bucket.
3. **Packet Transmission**: A packet is transmitted only if a token is available in the bucket. If the bucket is full, incoming packets are **discarded or buffered**.

#### Diagram: Leaky Bucket Working
```
   Packets In ----> [Bucket] ----> Packets Out (Constant Rate)
                   (Tokens Added at Fixed Rate)
```

---

## TCP Congestion Control

### Basic Idea
TCP uses a **dynamic window adjustment** mechanism to manage congestion, consisting of two main phases:

1. **Slow Start**
   - The **congestion window (cwnd)** starts small and **exponentially increases** until a threshold is reached or congestion is detected.
   - ```
     cwnd = cwnd * 2 (per RTT)
     ```

2. **Congestion Avoidance**
   - After reaching the threshold, the window size **increases linearly** to probe for available bandwidth.
   - ```
     cwnd = cwnd + 1 (per RTT)
     ```

3. **Congestion Detection**
   - If congestion is detected (e.g., packet loss), TCP **reduces the window size** and adjusts the threshold.

#### Diagram: TCP Congestion Behavior
```
   Window Size
     ^
     |   / (Exponential Growth - Slow Start)
     |  /
     | /
     |---- (Linear Growth - Congestion Avoidance)
     +-----------------------------------> Time
```

---

## Congestion Control in Frame Relay

### Basic Explanation
Frame Relay networks use **explicit congestion notification (ECN)** to manage congestion:
- **Forward Explicit Congestion Notification (FECN)**: Notifies the **destination** about congestion in the network.
- **Backward Explicit Congestion Notification (BECN)**: Notifies the **source** to reduce its transmission rate.
- **Discard Eligibility (DE) Bit**: Marks packets that can be discarded during congestion.

---

## Summary Points
- **Congestion control** is essential for maintaining **network performance** and preventing **packet loss** and **delay**.
- **Open loop** methods are **preventive**, while **closed loop** methods are **reactive**.
- The **Leaky Bucket Algorithm** smooths out bursty traffic to prevent congestion.
- **TCP Congestion Control** dynamically adjusts the transmission rate using **slow start** and **congestion avoidance**.
- **Frame Relay** uses **FECN, BECN, and DE bits** for congestion management.

# TCP vs UDP

## Comparison Table


| Feature               | TCP (Transmission Control Protocol)                          | UDP (User Datagram Protocol)                                |
|-----------------------|---------------------------------------------------------------|-------------------------------------------------------------|
| **Connection Type**   | Connection-oriented (requires handshake)                      | Connectionless (no handshake)                               |
| **Reliability**       | Reliable (guarantees delivery, order, and error correction)  | Unreliable (no guarantees)                                  |
| **Speed**             | Slower (due to overhead for reliability)                     | Faster (minimal overhead)                                   |
| **Error Control**     | Yes (checksum, acknowledgments, retransmissions)              | Limited (checksum only)                                     |
| **Flow Control**      | Yes (sliding window mechanism)                                | No                                                          |
| **Use Cases**         | Web browsing (HTTP/HTTPS), email (SMTP), file transfer (FTP) | Video streaming, VoIP, online gaming, DNS, DHCP            |
| **Overhead**          | High (20–60 bytes header)                                     | Low (8 bytes header)                                        |
| **Congestion Control**| Yes (adjusts transmission rate)                              | No                                                          |

---

## Key Differences Explained

### Connection Type
- **TCP**: Establishes a **connection** before data transfer using a **three-way handshake**. Ensures both parties are ready to communicate.
- **UDP**: Sends data **without establishing a connection**, making it faster but less reliable.

### Reliability
- **TCP**: Guarantees **delivery, order, and error-free transmission** using acknowledgments and retransmissions.
- **UDP**: **No guarantees**. Data may be lost, duplicated, or arrive out of order.

### Speed
- **TCP**: Slower due to **overhead** for reliability mechanisms (e.g., acknowledgments, flow control).
- **UDP**: Faster because it **skips reliability checks** and transmits data immediately.

### Error Control
- **TCP**: Uses **checksums, acknowledgments, and retransmissions** to ensure data integrity.
- **UDP**: Only uses a **checksum** for basic error detection. Corrupted packets are simply discarded.

### Flow Control
- **TCP**: Adjusts the **transmission rate** based on the receiver’s capacity using a **sliding window**.
- **UDP**: **No flow control**. The sender transmits data at its own pace, regardless of the receiver’s capacity.

### Overhead
- **TCP**: Larger header (20–60 bytes) due to additional fields for reliability and control.
- **UDP**: Smaller header (8 bytes), resulting in lower overhead.

---

## Use Cases

### When to Use TCP
- **Applications requiring reliability**: Web browsing (HTTP/HTTPS), email (SMTP), file transfer (FTP), and database transactions.
- **Scenario**: When every bit of data must arrive **intact and in order**, such as downloading a file or loading a webpage.

### When to Use UDP
- **Applications requiring speed and real-time performance**: Video streaming, VoIP, online gaming, and live broadcasts.
- **Scenario**: When **speed is more important than reliability**, such as in live video calls or multiplayer games, where occasional packet loss is acceptable.

---

## Summary Points
- **TCP** is ideal for applications where **reliability and ordered delivery** are critical.
- **UDP** is preferred for **real-time applications** where speed and low latency are more important than perfect reliability.
- **TCP** has higher overhead due to its reliability mechanisms, while **UDP** is lightweight and fast.
- Choose **TCP** for web browsing, file transfers, and emails; choose **UDP** for video calls, gaming, and live streaming.
