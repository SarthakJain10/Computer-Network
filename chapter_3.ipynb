# Circuit Switching

## Introduction
- **Circuit Switching** is a switching method where a **dedicated communication path** (circuit) is established between two devices before data transfer begins.
- Once the circuit is set up, the entire bandwidth is reserved for the duration of the communication.
- Used in traditional **telephone networks** and some legacy data networks.

---

## Working Phases

### 1. Setup Phase
- **Purpose**: Establish a dedicated end-to-end path.
- **Steps**:
  1. The **source device** sends a **request** to the network to set up a connection.
  2. The network **reserves resources** (bandwidth, switches, links) along the path.
  3. The **destination device** is notified, and if it accepts, the circuit is established.
- **Key Point**: Resources are **reserved** for the entire session, even if not used continuously.

### 2. Data Transfer Phase
- **Purpose**: Transmit data over the dedicated path.
- **Steps**:
  1. Data is sent as a **continuous stream** over the reserved path.
  2. No routing decisions are needed during transmission.
  3. The path remains **exclusively allocated** until the session ends.
- **Key Point**: Guaranteed **constant bandwidth** and **low latency** during transmission.

### 3. Teardown Phase
- **Purpose**: Release the reserved resources.
- **Steps**:
  1. Either the **source** or **destination** sends a **termination request**.
  2. The network **releases** the reserved resources (bandwidth, switches).
  3. The circuit is **disconnected**, and resources become available for other sessions.
- **Key Point**: Ensures efficient reuse of network resources.

---

## ASCII Diagram: Three Phases

```
Source Device ----------------------- Network ----------------------- Destination Device
       |                                   |
       |------ Setup Phase (Reserve Path) ------|
       |                                   |
       |===== Data Transfer Phase (Dedicated) =====|
       |                                   |
       |------ Teardown Phase (Release Path) -----|
```

---

## Delay Analysis

### 1. Setup Delay
- **Definition**: Time taken to establish the circuit.
- **Components**:
  - **Propagation delay**: Time for the setup request to travel through the network.
  - **Processing delay**: Time for switches/routers to process the request.
  - **Queuing delay**: Time spent waiting for resources to be allocated.

### 2. Transmission Delay
- **Definition**: Time taken to transmit data over the established path.
- **Components**:
  - **Propagation delay**: Time for data to travel from source to destination.
  - **Transmission time**: Time to push all data bits into the network.

---

## Efficiency

### Bandwidth Utilization
- **Advantages**:
  - **Guaranteed bandwidth**: No competition for resources during transmission.
  - **Low latency**: Ideal for real-time applications (e.g., voice calls).
- **Disadvantages**:
  - **Inefficient for bursty traffic**: Bandwidth is wasted if the circuit is idle.
  - **Resource blocking**: Other users cannot use the reserved path, even if it’s underutilized.

---

## Real-World Implementation: Telephone Networks
- **How it works**:
  - When you dial a number, the **setup phase** establishes a dedicated path.
  - During the call, the **data transfer phase** ensures clear, uninterrupted voice transmission.
  - After hanging up, the **teardown phase** releases the circuit.
- **Why it’s used**:
  - Guarantees **real-time communication** with minimal delay.
  - Ensures **consistent quality** for voice calls.

---

## Advantages and Disadvantages

| **Advantages**                          | **Disadvantages**                          |
|------------------------------------------|---------------------------------------------|
| Guaranteed bandwidth                     | Inefficient for bursty or intermittent data |
| Low latency                              | High setup delay                            |
| Suitable for real-time applications      | Resource wastage if circuit is idle         |
| Simple and predictable performance       | Not scalable for large networks            |

---

## Summary Points
- Circuit switching establishes a **dedicated path** for communication.
- Three phases: **Setup**, **Data Transfer**, and **Teardown**.
- **Efficiency**: High for continuous data, low for bursty traffic.
- **Delay**: Setup delay is significant, but transmission delay is minimal.
- **Real-world use**: Traditional telephone networks.

# Packet Switching

## Introduction
- **Packet Switching** is a method of transmitting data in small units called **packets** over a network.
- Each packet contains:
  - **Header** (source/destination address, sequence number, error control info)
  - **Payload** (actual data)
- Two main approaches:
  - **Datagram Network**
  - **Virtual Circuit Network**

---

## Datagram Network

### Definition
- In a **datagram network**, each packet is treated as an independent entity.
- Packets from the same source may take different paths to reach the destination.

### Working Mechanism
1. **No Path Setup**: No dedicated path is established before sending data.
2. **Packet Routing**: Each packet is routed independently using the **destination address** in its header.
3. **Routing Tables**: Routers use **routing tables** to determine the next hop for each packet.
4. **Delivery**: Packets may arrive out of order or be lost; reassembly and error handling are managed by the destination.

### Role of Routing Table
- Stores information about possible routes to various destinations.
- Updated dynamically using routing protocols (e.g., OSPF, BGP).
- Helps routers decide the best path for each packet.

### Efficiency and Delay
- **Efficiency**: High for bursty traffic; no resource reservation.
- **Delay**:
  - **Minimal setup delay** (no path establishment).
  - **Variable transmission delay** (depends on network congestion and routing path).

### Datagram Network in the Internet
- The **Internet** uses datagram packet switching (e.g., IP protocol).
- Packets are forwarded independently, making it robust and scalable.

---

## Virtual Circuit Network

### Definition
- A **virtual circuit network** establishes a logical path between source and destination before data transfer.
- All packets follow the same path, ensuring ordered delivery.

### Addressing
- Uses **Virtual Circuit Identifier (VCI)** in the packet header instead of full destination address.
- VCIs are unique for each virtual circuit.

### Three Phases

#### 1. Setup Phase
- A **setup request** is sent from source to destination.
- Routers along the path allocate resources and assign a VCI.
- Destination acknowledges the setup, establishing the virtual circuit.

#### 2. Data Transfer Phase
- Packets are sent along the established path using the VCI.
- Routers forward packets based on the VCI, ensuring ordered delivery.

#### 3. Teardown Phase
- Either end can send a **teardown request** to release resources.
- Routers free up allocated resources, and the virtual circuit is terminated.

### Working Mechanism
- **Path Establishment**: Logical path is set up before data transfer.
- **Ordered Delivery**: All packets follow the same path, arriving in order.
- **Resource Reservation**: Bandwidth and buffers are reserved for the duration of the circuit.

### Efficiency and Delay
- **Efficiency**: High for continuous data transfer; resources are reserved.
- **Delay**:
  - **Setup delay** (time to establish the virtual circuit).
  - **Low transmission delay** (dedicated path ensures minimal congestion).

### Use in WAN Technologies
- Used in **Frame Relay**, **ATM (Asynchronous Transfer Mode)**, and **MPLS (Multiprotocol Label Switching)**.
- Ideal for **real-time applications** (e.g., video conferencing, VoIP).

---

## ASCII Diagrams

### Packet Flow in Datagram Network
```
Source ---(Packet 1)---> Router A ---(Packet 1)---> Destination
          (Packet 2)---> Router B ---(Packet 2)--->
```

### Virtual Circuit Path Establishment
```
Source ---(Setup Request)---> Router A ---(Setup Request)---> Destination
<---(Ack)--- <---(Ack)---
Source ---(Data Packets)---> Router A ---(Data Packets)---> Destination
```

---

## Comparison: Datagram vs Virtual Circuit

| **Feature**            | **Datagram Network**                          | **Virtual Circuit Network**                  |
|------------------------|-----------------------------------------------|----------------------------------------------|
| **Path Setup**         | No setup; packets routed independently        | Path established before data transfer       |
| **Routing**            | Dynamic; each packet routed separately        | Fixed path for all packets                   |
| **Delay**              | Minimal setup delay; variable transmission    | Setup delay; low transmission delay          |
| **Reliability**        | No guaranteed delivery or order              | Guaranteed delivery and ordered packets       |
| **Order of Delivery**  | Packets may arrive out of order              | Packets arrive in order                      |
| **Resource Usage**     | Efficient for bursty traffic                  | Reserves resources; efficient for continuous data |

---

## Summary Points
- **Datagram Network**:
  - No path setup; packets routed independently.
  - Efficient for bursty traffic; used in the Internet (IP).
  - No guaranteed order or delivery.

- **Virtual Circuit Network**:
  - Path established before data transfer.
  - Guaranteed order and delivery; used in WAN technologies (Frame Relay, ATM).
  - Ideal for real-time, continuous data transfer.

# Message Switching

## Introduction
- **Message Switching** is a store-and-forward switching technique where **entire messages** are transmitted from source to destination via intermediate nodes.
- Each node **stores** the complete message, then **forwards** it to the next node.
- Used in early telecommunication systems and some modern applications like email.

---

## Working Mechanism
1. **Store**: The source sends the entire message to the first switch, which stores it temporarily.
2. **Forward**: The switch finds the best path to the next node and forwards the message.
3. **Repeat**: This process repeats at each intermediate node until the message reaches the destination.
4. **Delivery**: The destination receives the complete message.

---

## Delay Analysis
- **Store-and-Forward Delay**: Time taken to receive, store, and forward the entire message at each node.
- **Queuing Delay**: If the node is busy, messages wait in a queue.
- **Transmission Delay**: Time to send the message from one node to the next.
- **Total Delay**: Cumulative delay at each hop, which can be significant for long messages or busy networks.

---

## Advantages and Disadvantages

| **Advantages**                          | **Disadvantages**                          |
|------------------------------------------|---------------------------------------------|
| No need for dedicated paths              | High delay due to store-and-forward        |
| Efficient for variable-length messages   | Requires large storage at each node        |
| Simple implementation                    | Not suitable for real-time communication   |
| Can handle different data formats        | Lower throughput compared to packet switching |

---

# Comparison of Switching Techniques

## Detailed Comparison Table

| **Parameter**          | **Circuit Switching**                          | **Packet Switching**                          | **Message Switching**                        |
|------------------------|-----------------------------------------------|-----------------------------------------------|---------------------------------------------|
| **Setup Requirement**  | Dedicated path setup required                 | No setup (datagram) or logical path (virtual circuit) | No dedicated path setup                     |
| **Delay**              | Setup delay + minimal transmission delay      | Minimal setup delay + variable transmission delay | High delay (store-and-forward at each node) |
| **Efficiency**         | High for continuous data; inefficient for bursty traffic | High for bursty traffic; efficient bandwidth use | Low efficiency; high overhead              |
| **Reliability**        | High (dedicated path)                         | Variable (depends on protocol)               | Moderate (message integrity preserved)     |
| **Complexity**         | Low (simple, predictable)                     | High (routing, reassembly, error handling)   | Moderate (storage, forwarding logic)       |
| **Use Cases**           | Real-time applications (voice, video calls)  | Internet, bursty data (web, email)           | Early telecom, email, non-real-time data    |

---

## Key Observations
- **Circuit Switching**: Best for **real-time, continuous** data (e.g., telephone networks).
- **Packet Switching**: Best for **bursty, variable-length** data (e.g., Internet, email).
- **Message Switching**: Suitable for **non-real-time, complete message** transfer (e.g., early email systems).

---

## Summary Points
- **Message Switching** is a **store-and-forward** technique, ideal for non-real-time, complete message transfer but suffers from high delay.
- **Circuit Switching** provides **dedicated paths**, ensuring low latency and high reliability for continuous data.
- **Packet Switching** is **flexible and efficient**, making it the backbone of modern networks like the Internet.
- Choose the switching technique based on **data type, delay tolerance, and reliability requirements**.

# Switching Fabrics

## Introduction
- **Switching Fabric** refers to the **interconnection network** inside a switch or router that moves data between input and output ports.
- It is the **core component** that determines how data is forwarded within a switching system.
- The primary purpose is to **efficiently route** data from source to destination with minimal delay and congestion.

---

## Types Overview
Switching fabrics can be categorized into three main types:

1. **Space Division Switching**
   - Uses physical paths to connect inputs to outputs.
   - Example: Crossbar switch.

2. **Time Division Switching**
   - Uses time slots to share a single physical path among multiple inputs/outputs.
   - Example: TSI (Time-Slot Interchange).

3. **Hybrid Switching**
   - Combines space and time division techniques for improved performance.

---

# Space Division Switching

## Introduction
- **Space Division Switching** establishes a **dedicated physical path** between input and output ports for the duration of a connection.
- Each connection gets its own path, ensuring **no interference** with other connections.
- Commonly used in **circuit-switched networks** like traditional telephone systems.

---

## Working
### Crossbar Switch
- A **crossbar switch** is a grid of horizontal (input) and vertical (output) lines.
- At each intersection, a **crosspoint** can be closed to establish a connection.
- Only one crosspoint is closed at a time for a given input or output.

#### How it Works:
1. **Request**: An input line requests a connection to an output line.
2. **Path Establishment**: The switch closes the crosspoint, creating a dedicated path.
3. **Data Transfer**: Data flows directly from input to output without interference.
4. **Release**: The crosspoint is opened, releasing the path.

---

## Switching Matrix
- A **switching matrix** is a grid of crosspoints arranged in rows (inputs) and columns (outputs).
- Each crosspoint can be **electronically or mechanically** controlled.

---

## Control Mechanism
- The control unit manages the **closing and opening** of crosspoints.
- It ensures **no two inputs** are connected to the same output simultaneously (prevents blocking).
- Uses algorithms to **optimize path selection** and minimize blocking probability.

---

## ASCII Diagram: Crossbar Switching Matrix

```
   O1  O2  O3
I1 [x] [ ] [ ]
I2 [ ] [x] [ ]
I3 [ ] [ ] [x]
```
- `[x]` = Closed crosspoint (active connection)
- `[ ]` = Open crosspoint (no connection)

---

## Advantages and Disadvantages

| **Advantages**                          | **Disadvantages**                          |
|------------------------------------------|---------------------------------------------|
| Dedicated, interference-free paths       | **Blocking**: Not all inputs can connect to outputs simultaneously |
| Low latency for established connections  | **Cost**: Complex and expensive for large-scale systems |
| Simple and predictable performance       | Scalability issues due to physical crosspoints |

---

## Summary Points
- **Switching Fabric** is the backbone of any switching system, determining how data is routed.
- **Space Division Switching** uses physical paths (e.g., crossbar switch) for dedicated connections.
- **Crossbar Switch** provides direct input-to-output connections but suffers from blocking and scalability issues.
- Ideal for **real-time, circuit-switched** applications like traditional telephony.

# Time Division Switching

## Introduction
- **Time Division Switching** allows multiple inputs to share a single physical path by dividing the transmission time into **discrete time slots**.
- Each input is assigned a specific time slot, enabling efficient utilization of the transmission medium.
- Commonly used in **digital telephony** and **data networks**.

---

## Working Mechanism
### Time Slot Interchange (TSI)
- **TSI** is a technique where time slots from input frames are **rearranged** and placed into output frames.
- Each input frame contains multiple time slots, and the switch **maps** these slots to the desired output slots.

#### How it Works:
1. **Input Framing**: Data arrives in fixed-size frames, each divided into time slots.
2. **Slot Mapping**: The switch reads the input slots and writes them into the correct output slots based on a predefined schedule.
3. **Output Framing**: The rearranged slots form the output frame, which is sent to the destination.

---

## ASCII Diagram: Time Slot Switching

```
Input Frame:  [Slot1][Slot2][Slot3][Slot4]
              |      |      |      |
Switch:       |------|------|------|
              v      v      v      v
Output Frame: [Slot3][Slot1][Slot4][Slot2]
```

---

## Advantages and Disadvantages

| **Advantages**                          | **Disadvantages**                          |
|------------------------------------------|---------------------------------------------|
| Efficient use of bandwidth              | Requires precise synchronization            |
| Supports multiple connections on a single path | Complex scheduling and slot management      |
| Low hardware complexity                  | Limited scalability for very high-speed networks |

---

# Multiple-Stage Switching

## Introduction
- **Multiple-Stage Switching** uses a network of smaller switches arranged in stages to overcome the limitations of single-stage switches (e.g., blocking, scalability).
- Each stage consists of multiple crossbar or TSI switches, connected in a specific pattern.

---

## Design Concept
- **Need for Multi-Stage Design**:
  - Single-stage switches suffer from **blocking** (unable to connect all inputs to outputs simultaneously).
  - Multi-stage designs improve **scalability** and **reduce blocking probability**.

- **Working Concept**:
  - Data passes through multiple stages of smaller switches.
  - Each stage routes data closer to its destination, reducing the chance of blocking.

---

## Types
- **Blocking Networks**: Some input-output connections may be blocked due to conflicts in intermediate stages.
- **Non-Blocking Networks**: Designed to ensure that any input can connect to any free output, regardless of other connections.

---

## ASCII Diagram: Multi-Stage Switching Structure

```
Stage 1: [Switch A] --- [Switch B] --- [Switch C]
           |              |              |
Stage 2: [Switch D] --- [Switch E] --- [Switch F]
           |              |              |
Stage 3: [Switch G] --- [Switch H] --- [Switch I]
```

---

# Combined Switching: Time-Space-Time (TST) and Space-Time-Space (STS)

## Working
### Time-Space-Time (TST) Switching
1. **Time Stage**: Input time slots are rearranged.
2. **Space Stage**: The rearranged slots are routed through a space division switch (e.g., crossbar).
3. **Time Stage**: Output time slots are rearranged again before transmission.

### Space-Time-Space (STS) Switching
1. **Space Stage**: Inputs are routed through a space division switch.
2. **Time Stage**: Time slots are rearranged.
3. **Space Stage**: The rearranged slots are routed to the final outputs.

---

## Benefits
| **Benefit**                          | **Description**                                      |
|--------------------------------------|------------------------------------------------------|
| **Reduced Blocking**                 | Combines advantages of both time and space division |
| **Scalability**                      | Easier to expand compared to single-stage switches  |
| **Flexibility**                      | Supports both time-division and space-division routing |

---

## Summary Points
- **Time Division Switching** uses time slots to share a single path efficiently, ideal for digital networks.
- **Multiple-Stage Switching** improves scalability and reduces blocking by using smaller switches in stages.
- **Combined Switching (TST/STS)** leverages both time and space division for optimal performance and flexibility.
- These techniques are foundational in modern **telecommunication networks** and **data switching systems**.

# Multiplexing

## Introduction
- **Multiplexing** is the process of combining multiple signals into one shared medium for efficient transmission.
- Enables multiple users/devices to share a single communication channel.
- Two primary types: **Frequency Division Multiplexing (FDM)** and **Time Division Multiplexing (TDM)**.

---

## Types

### 1. Frequency Division Multiplexing (FDM)
- **Definition**: Divides the total bandwidth of a channel into **non-overlapping frequency bands**.
- **Working Mechanism**:
  - Each signal is assigned a unique frequency band.
  - Signals are transmitted simultaneously without interference.
- **Example**: Radio broadcasting, cable TV.

### 2. Time Division Multiplexing (TDM)
- **Definition**: Divides the transmission time into **time slots**.
- **Working Mechanism**:
  - Each signal gets a specific time slot for transmission.
  - Slots are repeated in cycles, allowing multiple signals to share the channel.
- **Example**: Digital telephony, ISDN.

---

## Working

### FDM vs TDM: Key Differences

| **Parameter**       | **FDM**                                      | **TDM**                                      |
|---------------------|-----------------------------------------------|-----------------------------------------------|
| **Division Basis**  | Frequency                                     | Time                                         |
| **Bandwidth Usage** | Each signal uses a fixed frequency band      | Each signal uses the full bandwidth for a short time |
| **Complexity**      | Requires filters to separate frequencies     | Requires precise synchronization for time slots |
| **Use Case**        | Analog signals (radio, TV)                   | Digital signals (telephony, data networks)   |

---

# T1 and E1 Standards

## Explanation

### T1 Standard
- **Definition**: A digital transmission standard used in North America and Japan.
- **Data Rate**: 1.544 Mbps.
- **Structure**:
  - 24 channels, each with 8-bit samples.
  - Uses **TDM** for multiplexing.
  - Each channel carries 64 Kbps (8 bits × 8,000 samples/sec).

### E1 Standard
- **Definition**: A digital transmission standard used in Europe and most of the world.
- **Data Rate**: 2.048 Mbps.
- **Structure**:
  - 32 channels, each with 8-bit samples.
  - Uses **TDM** for multiplexing.
  - 30 channels for data, 1 for framing, and 1 for signaling.

---

## Comparison: T1 vs E1

| **Parameter**       | **T1**                          | **E1**                          |
|---------------------|----------------------------------|----------------------------------|
| **Data Rate**       | 1.544 Mbps                      | 2.048 Mbps                      |
| **Channels**        | 24 channels                     | 32 channels (30 usable)          |
| **Geographical Use**| North America, Japan            | Europe, most of the world        |
| **Framing**         | Uses 1 framing bit per 193 bits | Uses 1 channel for framing       |

---

# Design Examples

## Explanation

### Simple Switching System Design
- **Objective**: Design a system to switch multiple input signals to multiple outputs.
- **Components**:
  - **Input Ports**: Receive signals from users/devices.
  - **Switching Fabric**: Routes signals to the correct output (e.g., crossbar or TSI).
  - **Output Ports**: Transmit signals to destinations.

#### Example: Crossbar Switch Design
- **Inputs**: 4 users.
- **Outputs**: 4 destinations.
- **Switching Fabric**: 4x4 crossbar matrix.
- **Working**:
  - Each user connects to an input port.
  - The crossbar switch establishes a direct path from input to output.

---

### Multiplexing in Switching Systems
- **Objective**: Combine multiple signals for transmission over a shared medium.
- **Example**: TDM in Digital Telephony
  - Multiple voice calls are sampled and assigned time slots.
  - The multiplexed signal is transmitted over a single line.
  - At the receiving end, signals are demultiplexed and routed to the correct destination.

---

## Summary Points
- **Multiplexing** enables efficient use of bandwidth by combining multiple signals.
- **FDM** divides bandwidth by frequency, while **TDM** divides by time.
- **T1** and **E1** are digital transmission standards using TDM, differing in data rates and geographical use.
- **Switching and multiplexing** are applied in real-world systems like telephony and data networks to optimize resource usage.
