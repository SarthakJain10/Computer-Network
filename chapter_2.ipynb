# 1. Framing

## Introduction
- **Framing** is the process of dividing the stream of bits received from the network layer into manageable units called **frames**.
- It is a fundamental function of the **Data Link Layer (DLL)**.

## Key Concepts
- **Frame**: A unit of data that includes:
  - **Header** (source/destination addresses, control info)
  - **Payload** (actual data)
  - **Trailer** (error detection bits, e.g., CRC)
- **Purpose of Framing**:
  - Enables error detection and correction.
  - Facilitates flow control.
  - Provides synchronization between sender and receiver.

## Detailed Explanation
- **Why Framing is Needed**:
  - The physical layer transmits a raw bit stream. Framing helps the receiver identify the start and end of each frame.
  - Ensures data integrity and orderly transmission.
- **Types of Framing Techniques**:
  1. **Fixed Size Framing**
  2. **Variable Size Framing**

---

# 2. Fixed Size Framing

## Introduction
- In **fixed size framing**, all frames are of the same, predetermined length.

## Key Concepts
- **Frame Size**: Constant and agreed upon by sender and receiver.
- **Example**: ATM (Asynchronous Transfer Mode) uses fixed-size cells of 53 bytes.

## Working Mechanism
1. The sender divides data into fixed-size chunks.
2. Each chunk is encapsulated into a frame with a header and trailer.
3. The receiver reassembles the data based on the fixed frame size.

## Advantages
- Simple to implement.
- No need for delimiters or special characters.
- Efficient for real-time applications (e.g., voice/video).

## Disadvantages
- **Padding** may be required if data doesn’t fit exactly, leading to overhead.
- Inflexible for variable data sizes.

---

# 3. Variable Size Framing

## Introduction
- In **variable size framing**, frames can vary in length.

## Key Concepts
- **Need for Variable Framing**:
  - Accommodates varying data sizes efficiently.
  - Reduces overhead by avoiding padding.

## Techniques Used
1. **Character/Byte Count**:
   - A field in the header specifies the frame length.
2. **Byte Stuffing**:
   - Uses special characters to mark frame boundaries.
3. **Bit Stuffing**:
   - Inserts extra bits to mark frame boundaries.

## Advantages
- No padding overhead.
- Flexible for different data sizes.

## Disadvantages
- More complex to implement.
- Requires delimiters or special characters.

---

# 4. Bit Stuffing

## Introduction
- **Bit Stuffing** is a technique used to ensure that the frame boundary (flag) is not mistakenly identified within the data.

## Key Concepts
- **Flag**: A special bit pattern (e.g., `01111110`) marks the start/end of a frame.
- **Stuffing**: Inserts a `0` after every five consecutive `1`s in the data to prevent false flags.

## Working Mechanism
1. The sender scans the data for five consecutive `1`s.
2. If found, a `0` is inserted after the fifth `1`.
3. The receiver removes the stuffed `0` after five `1`s.

## Example
- **Original Data**: `11111011111`
- **After Stuffing**: `111110111110`

## Advantages
- Ensures transparent data transmission.
- Prevents false frame boundaries.

## Limitations
- Overhead due to extra bits.

---

# 5. Byte Stuffing

## Introduction
- **Byte Stuffing** is similar to bit stuffing but operates at the byte level.

## Key Concepts
- **Flag Byte**: Marks frame boundaries (e.g., `ESC` or `FLAG`).
- **Escape Byte**: Indicates that the next byte is data, not a flag.

## Working Mechanism
1. The sender inserts an `ESC` byte before any `FLAG` or `ESC` in the data.
2. The receiver removes the `ESC` byte during processing.

## Example
- **Original Data**: `A FLAG B ESC C`
- **After Stuffing**: `A ESC FLAG B ESC ESC C`

## Comparison with Bit Stuffing
| Feature        | Bit Stuffing               | Byte Stuffing               |
|----------------|----------------------------|-----------------------------|
| **Unit**       | Bits                        | Bytes                       |
| **Overhead**   | Lower (single bits)         | Higher (extra bytes)        |
| **Complexity** | Lower                      | Higher                      |

---

# 6. Flow Control

## Introduction
- **Flow Control** manages the rate of data transmission to prevent overwhelming the receiver.

## Key Concepts
- **Need for Flow Control**:
  - Prevents data loss due to buffer overflow.
  - Ensures the receiver can process incoming data.

## Techniques
1. **Stop-and-Wait**:
   - Sender waits for acknowledgment (ACK) after each frame.
2. **Sliding Window**:
   - Allows multiple frames to be sent before receiving an ACK.

## Working Explanation
- **Stop-and-Wait**:
  1. Sender transmits a frame.
  2. Waits for ACK.
  3. Proceeds only after receiving ACK.
- **Sliding Window**:
  1. Sender maintains a "window" of frames that can be sent without ACK.
  2. Receiver acknowledges frames, and the window slides forward.

## Advantages and Limitations
| Technique      | Advantages                     | Limitations                     |
|----------------|---------------------------------|---------------------------------|
| Stop-and-Wait  | Simple, no buffer needed        | Inefficient for long distances  |
| Sliding Window | Efficient, high throughput      | Requires buffering              |

---

# 7. Error Control

## Introduction
- **Error Control** ensures data integrity by detecting and correcting errors during transmission.

## Key Concepts
- **Types of Errors**:
  - **Single-bit**: One bit is flipped.
  - **Burst**: Multiple consecutive bits are corrupted.

## Techniques
1. **Error Detection**:
   - **Parity Check**: Adds a parity bit to detect single-bit errors.
   - **CRC (Cyclic Redundancy Check)**: Detects burst errors.
2. **Error Correction**:
   - **Hamming Code**: Corrects single-bit errors.
3. **ARQ Protocols**:
   - **Stop-and-Wait ARQ**: Retransmits lost/corrupted frames.
   - **Go-Back-N**: Retransmits all frames from the error point.
   - **Selective Repeat**: Retransmits only the corrupted frames.

## Working Explanation
- **Stop-and-Wait ARQ**:
  1. Sender transmits a frame.
  2. Waits for ACK or NAK (Negative Acknowledgment).
  3. Retransmits if NAK is received.

# Data Link Layer Protocols

# 1. Protocols for Noiseless Channel

---

## (a) Simplest Protocol

### Introduction
- The **Simplest Protocol** assumes an **ideal, noiseless channel** with no errors or flow control issues.
- Used as a baseline to understand more complex protocols.

### Key Concepts
- **Assumptions**:
  - No transmission errors.
  - Infinite receiver buffer.
  - No need for flow or error control.

### Working Mechanism
1. **Sender**:
   - Sends frames continuously without waiting.
   - No acknowledgments (ACKs) or timers.
2. **Receiver**:
   - Accepts all incoming frames without feedback.

### Diagram
```
Sender --------------------------> Receiver
   Frame1   Frame2   Frame3   ...
```

### Advantages and Disadvantages
- **Advantages**:
  - Extremely simple.
  - No overhead for ACKs or timers.
- **Disadvantages**:
  - Unrealistic for real-world channels.
  - No error or flow control.

### Summary Points
- Only theoretical; not practical for real networks.
- Serves as a foundation for understanding more realistic protocols.

---

## (b) Stop-and-Wait Protocol

### Introduction
- **Stop-and-Wait Protocol** introduces **flow control** for noiseless channels.
- Ensures the receiver can handle incoming frames.

### Key Concepts
- **Flow Control**: Prevents overwhelming the receiver.
- **ACK (Acknowledgment)**: Receiver sends ACK after processing each frame.

### Working Mechanism
1. **Sender**:
   - Sends **one frame** and waits for ACK.
   - Only sends the next frame after receiving ACK.
2. **Receiver**:
   - Processes the frame and sends ACK.

### Step-by-Step Explanation
1. Sender transmits Frame 0.
2. Receiver receives Frame 0 and sends ACK 0.
3. Sender receives ACK 0 and transmits Frame 1.
4. Repeat for each frame.

### Diagram
```
Sender --------------------------> Receiver
   Frame0 --------------------->
        <----------------------- ACK0
   Frame1 --------------------->
        <----------------------- ACK1
```

### Advantages and Limitations
- **Advantages**:
  - Simple flow control.
  - No buffer required at the receiver.
- **Limitations**:
  - **Inefficient**: Sender waits idle after each frame.
  - Not suitable for long-distance or noisy channels.

---

# 2. Protocols for Noisy Channel

---

## (a) Stop-and-Wait ARQ

### Introduction
- **Stop-and-Wait ARQ (Automatic Repeat reQuest)** adds **error control** to the Stop-and-Wait protocol.
- Designed for **noisy channels** where frames can be lost or corrupted.

### Key Concepts
- **ACK/NAK**: Receiver sends ACK for correct frames, NAK for errors.
- **Timer**: Sender uses a timer to detect lost frames/ACKs.

### Working Mechanism
1. **Sender**:
   - Sends Frame 0 and starts a timer.
   - Waits for ACK 0.
   - If timer expires, retransmits Frame 0.
2. **Receiver**:
   - Sends ACK 0 if Frame 0 is correct.
   - Sends NAK 0 if Frame 0 is corrupted.

### Handling Lost Frames and ACKs
- **Lost Frame**: Timer expires → retransmit.
- **Lost ACK**: Sender retransmits; receiver discards duplicates using sequence numbers.

### Diagram
```
Sender --------------------------> Receiver
   Frame0 --------------------->
        <----------------------- ACK0 (or NAK0)
   Frame1 --------------------->
        <----------------------- ACK1 (or NAK1)
   (Timer expires) Frame0 ------>
```

### Advantages and Limitations
- **Advantages**:
  - Simple error and flow control.
- **Limitations**:
  - **Low efficiency**: Sender waits for ACK after each frame.
  - **Throughput**: Degrades with long propagation delays.

---

## (b) Go-Back-N ARQ

### Introduction
- **Go-Back-N ARQ** improves efficiency by allowing the sender to transmit **multiple frames** before receiving ACKs.
- Uses a **sliding window** for flow and error control.

### Key Concepts
- **Sliding Window**: Sender can send up to `N` frames without ACK.
- **Sequence Numbers**: Frames are numbered modulo `2^n`.
- **Cumulative ACK**: ACK for frame `n` implies all frames up to `n` are received correctly.

### Working Mechanism
1. **Sender**:
   - Sends frames 0 to `N-1` (window size).
   - Starts timer for the oldest unacknowledged frame.
   - If timer expires for frame `i`, retransmits frames `i` to `i+N-1`.
2. **Receiver**:
   - Discards out-of-order frames.
   - Sends ACK for the highest in-order frame received.

### Diagram
```
Sender Window: [0,1,2,3] -> [4,5,6,7] (slides forward on ACK)
Receiver Window: [0] -> [4] (after ACK 3)
```

### Advantages and Disadvantages
- **Advantages**:
  - Higher throughput than Stop-and-Wait.
  - Efficient for error-prone channels.
- **Disadvantages**:
  - **Retransmission Overhead**: Resends all frames after the lost/corrupted frame.
  - Requires buffering at the receiver.

---

## (c) Selective Repeat ARQ

### Introduction
- **Selective Repeat ARQ** further improves efficiency by retransmitting **only lost/corrupted frames**.
- Uses **individual ACKs** and buffering at the receiver.

### Key Concepts
- **Selective ACK**: Receiver acknowledges each frame individually.
- **Receiver Window**: Buffers out-of-order frames until missing frames arrive.

### Working Mechanism
1. **Sender**:
   - Sends frames within the window.
   - Retransmits only frames for which NAK is received.
2. **Receiver**:
   - Buffers out-of-order frames.
   - Sends ACK for each correctly received frame.

### Diagram
```
Sender Window: [0,1,2,3] -> [4,5,6,7]
Receiver Window: [0,1,3] (buffers 3, waits for 2)
   <----------------------- NAK 2
   Frame2 ---------------------->
```

### Advantages and Disadvantages
- **Advantages**:
  - Most efficient for noisy channels.
  - Minimizes retransmission overhead.
- **Disadvantages**:
  - Complex receiver buffering.
  - Requires more memory and processing.

# 1. Pure ALOHA

## Introduction
- **Pure ALOHA** is a simple, decentralized **random access protocol** for shared medium networks.
- Allows stations to transmit data whenever they have it, without checking the channel state.

## Key Concepts
- **Random Access**: Stations transmit without coordination.
- **Collisions**: Occur if two or more stations transmit simultaneously.
- **Throughput**: Fraction of successful transmissions over total attempts.

## Working Mechanism
1. Station transmits a frame immediately when it is ready.
2. If no ACK is received within a timeout, assume collision.
3. Station waits a random time and retransmits.

### Collision Scenario
- If two stations transmit at the same time, both frames are corrupted.

## Throughput and Efficiency
- Throughput: **S = G x e^-2G**.
- Maximum throughput: **18%** of channel capacity.
- Low efficiency due to high collision probability.

## Advantages and Disadvantages
- **Advantages**:
  - Simple and decentralized.
- **Disadvantages**:
  - Low throughput.
  - High collision probability.

## Diagram: ALOHA Transmission Timeline
```
Station A: |-----Frame A-----|
Station B:    |-----Frame B-----| (Collision)
```

---

# 2. Slotted ALOHA

## Introduction
- **Slotted ALOHA** improves Pure ALOHA by dividing time into discrete **slots**.
- Stations can only transmit at the beginning of a slot.

## Key Concepts
- **Time Slots**: Fixed duration, synchronized across stations.
- **Reduced Collisions**: Frames cannot overlap partially.

## Working Mechanism
1. Station waits for the next slot boundary.
2. Transmits the frame at the slot start.
3. If collision, waits a random number of slots and retransmits.

## Throughput Improvement
- Throughput: **S = G x e^-G**
- Maximum throughput: **36.8%** (double Pure ALOHA).

## Advantages and Disadvantages
- **Advantages**:
  - Higher throughput than Pure ALOHA.
- **Disadvantages**:
  - Requires global synchronization.

## Diagram: Slotted ALOHA Slots
```
Time: |Slot 1|Slot 2|Slot 3|...
      |FrameA|      |FrameB|
```

---

# 3. CSMA (Carrier Sense Multiple Access)

## Introduction
- **CSMA** improves ALOHA by **listening to the channel** before transmitting.
- Reduces collision probability.

## Key Concepts
- **Carrier Sensing**: Station checks if the channel is idle.
- **Types of CSMA**:
  - **1-persistent**: Transmit immediately if channel is idle.
  - **Non-persistent**: Wait random time if channel is busy.
  - **p-persistent**: Transmit with probability `p` if channel is idle.

## Working of Each Type
1. **1-persistent**:
   - If channel is idle, transmit; else, keep listening.
2. **Non-persistent**:
   - If channel is busy, wait random time and recheck.
3. **p-persistent**:
   - If channel is idle, transmit with probability `p`; else, wait next slot.

## Advantages and Limitations
- **Advantages**:
  - Reduces collisions compared to ALOHA.
- **Limitations**:
  - Still possible for collisions (propagation delay).

# 4. CSMA/CD (Collision Detection)

## Introduction
- **CSMA/CD** adds **collision detection** to CSMA.
- Used in wired networks (e.g., Ethernet).

## Key Concepts
- **Collision Detection**: Stations detect collisions while transmitting.
- **Jam Signal**: Sent to notify all stations of a collision.
- **Backoff Algorithm**: Random delay before retransmission.

## Working Mechanism
1. Station senses the channel.
2. If idle, transmits; else, waits.
3. If collision detected, sends jam signal and stops transmission.
4. Waits random time (backoff) and retransmits.

## Step-by-Step Explanation
1. **Transmission**: Station sends frame.
2. **Collision Detection**: Monitors channel during transmission.
3. **Jam Signal**: Sent for ~48 bits to ensure all stations detect collision.
4. **Backoff**: Waits random time using truncated binary exponential backoff.

## Advantages and Disadvantages
- **Advantages**:
  - Effective in wired networks.
- **Disadvantages**:
  - Inefficient for high-load or long-propagation-delay networks.

## Diagram: CSMA/CD Collision Scenario
```
Station A: |-----Frame A-----| Jam
Station B:    |-----Frame B-----| (Collision)
```

# 5. Ethernet (Detailed)

## Introduction
- **Ethernet** is the most widely used wired LAN technology.
- Defined by **IEEE 802.3** standard.

## Key Concepts
- **MAC Addressing**: Unique 48-bit address for each network interface.
- **CSMA/CD**: Used in traditional Ethernet for collision handling.

## Working Principle
- Stations share a common bus (historically) or switch.
- Uses **CSMA/CD** for media access (in half-duplex mode).

## Frame Structure
| Field               | Size (bytes) | Description                          |
|---------------------|--------------|--------------------------------------|
| **Preamble**        | 7            | Synchronization                      |
| **SFD**             | 1            | Start of frame delimiter             |
| **Destination MAC**  | 6            | Receiver's MAC address               |
| **Source MAC**       | 6            | Sender's MAC address                 |
| **Type/Length**      | 2            | Protocol type or frame length       |
| **Data**            | 46–1500      | Payload                              |
| **CRC**             | 4            | Error detection                      |

## MAC Addressing
- **MAC Address**: 6-byte unique identifier (e.g., `00:1A:2B:3C:4D:5E`).
- Used for frame forwarding in LANs.

## Role of CSMA/CD in Ethernet
- Historically used in half-duplex Ethernet.
- Modern Ethernet uses **full-duplex** with switches, eliminating collisions.

## Ethernet Operation in Real Networks
- **Switches**: Forward frames based on MAC addresses.
- **Full-Duplex**: Stations can transmit and receive simultaneously.

## Diagram: Ethernet Frame Structure
```
+---------------------+---------------------+---------------------+
| Preamble (7B)       | SFD (1B)            | Dest MAC (6B)       |
+---------------------+---------------------+---------------------+
| Source MAC (6B)     | Type/Length (2B)    | Data (46-1500B)     |
+---------------------+---------------------+---------------------+
| CRC (4B)            |                     |                     |
+---------------------+---------------------+---------------------+
```
