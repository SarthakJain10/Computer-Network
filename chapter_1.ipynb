# 1. Introduction to Computer Networking

## Introduction
- A **computer network** is a group of interconnected devices (computers, servers, printers, etc.) that can communicate and share resources.
- Networks enable data exchange, resource sharing, and communication.

## Key Concepts
- **Nodes**: Devices connected to the network (e.g., computers, printers).
- **Links**: Communication channels (wired/wireless) connecting nodes.
- **Protocols**: Rules for data exchange (e.g., TCP/IP, HTTP).

## Purpose and Importance
- **Resource Sharing**: Share hardware (printers, storage) and software.
- **Communication**: Enable email, video calls, and messaging.
- **Data Access**: Access remote databases or cloud services.
- **Cost Efficiency**: Reduce hardware/software duplication.

## Advantages
- **Convenience**: Access data from anywhere.
- **Reliability**: Backup and redundancy.
- **Scalability**: Easy to add new devices.

## Real-World Uses
- **Internet**: Global network of networks.
- **Office Networks**: File sharing and collaboration.
- **Online Banking**: Secure transactions.

# 2. Network Topologies

## Introduction
- **Topology** refers to the physical/logical arrangement of devices in a network.

## Key Concepts
- **Physical Topology**: How devices are physically connected.
- **Logical Topology**: How data flows between devices.

---

### Bus Topology
- **Structure**: All devices connected to a single cable (backbone).
- **Advantages**: Easy to install, cost-effective.
- **Disadvantages**: Single point of failure, slows down with more devices.

```
  [Device]----[Device]----[Device]----[Device]
```

### Star Topology
- **Structure**: All devices connected to a central hub/switch.
- **Advantages**: Easy to manage, scalable.
- **Disadvantages**: Hub failure disrupts the network.

```
      [Hub]
     / | | \
  [D1][D2][D3][D4]
```

### Ring Topology
- **Structure**: Devices connected in a closed loop.
- **Advantages**: Equal access, no collisions.
- **Disadvantages**: Single point of failure, difficult to troubleshoot.

```
  [D1] <---> [D2]
   ^            |
   |            v
  [D4] <---> [D3]
```

### Mesh Topology
- **Structure**: Every device connected to every other device.
- **Advantages**: High redundancy, reliable.
- **Disadvantages**: Expensive, complex setup.

```
  [D1]--[D2]--[D3]
   | \    |    /
   |  \  |  /
  [D4]--[D5]
```

### Tree Topology
- **Structure**: Hierarchical structure (star + bus).
- **Advantages**: Scalable, easy to manage.
- **Disadvantages**: Complex cabling, single point of failure at root.

```
      [Root]
     /      \
  [D1]     [D2]
   |        /  \
  [D3]   [D4] [D5]
```

# 3. Types of Area Networks

## Introduction
- Networks are classified based on their **geographical coverage**.

---

### LAN (Local Area Network)
- **Definition**: Covers a small area (e.g., home, office).
- **Coverage Area**: Up to a few kilometers.
- **Examples**: Home Wi-Fi, office network.

### MAN (Metropolitan Area Network)
- **Definition**: Covers a city or large campus.
- **Coverage Area**: Up to 50 km.
- **Examples**: City-wide Wi-Fi, university campus network.

### WAN (Wide Area Network)
- **Definition**: Covers a large geographical area.
- **Coverage Area**: Global (e.g., internet).
- **Examples**: Internet, multinational company networks.

# 4. Transmission Modes

## Introduction
- **Transmission mode** defines the direction of data flow between devices.

---

### Simplex
- **Definition**: Data flows in **one direction only**.
- **Direction**: Unidirectional.
- **Example**: TV broadcast, keyboard input.

### Half-Duplex
- **Definition**: Data flows in **both directions, but not simultaneously**.
- **Direction**: Bidirectional (one at a time).
- **Example**: Walkie-talkie.

### Full-Duplex
- **Definition**: Data flows in **both directions simultaneously**.
- **Direction**: Bidirectional (simultaneous).
- **Example**: Phone call, internet browsing.

# 5. Network Devices

## Introduction
- **Network devices** facilitate data transmission and network management.

---

### Hub
- **Definition**: Connects multiple devices in a LAN.
- **Function**: Broadcasts data to all connected devices.
- **Working**: No intelligence; forwards all data.
- **OSI Layer**: Physical Layer.
- **Example**: Ethernet hub.

### Repeater
- **Definition**: Amplifies signals to extend network range.
- **Function**: Regenerates weak signals.
- **Working**: Operates at the physical layer.
- **OSI Layer**: Physical Layer.
- **Example**: Signal booster.

### Bridge
- **Definition**: Connects two LANs or network segments.
- **Function**: Filters and forwards data based on MAC addresses.
- **Working**: Operates at the data link layer.
- **OSI Layer**: Data Link Layer.
- **Example**: Connecting two office networks.

### Switch
- **Definition**: Connects devices in a LAN and forwards data intelligently.
- **Function**: Uses MAC addresses to forward data to the correct device.
- **Working**: Operates at the data link layer.
- **OSI Layer**: Data Link Layer.
- **Example**: Ethernet switch.

### Router
- **Definition**: Connects multiple networks (LAN/WAN).
- **Function**: Routes data between networks using IP addresses.
- **Working**: Operates at the network layer.
- **OSI Layer**: Network Layer.
- **Example**: Home Wi-Fi router.

# 1. OSI Model (Open Systems Interconnection Model)

## Introduction
- The **OSI Model** is a conceptual framework used to standardize network functions into **7 layers**.
- **Purpose**: To enable communication between different systems without requiring changes to the underlying hardware or software.
- **Importance**: Provides a modular approach to networking, making troubleshooting, development, and standardization easier.

---

## Layer-wise Explanation

### 1. Physical Layer
- **Main Function**: Transmits raw **bit streams** over physical media.
- **Key Responsibilities**:
  - Defines physical characteristics (cables, connectors, voltage levels).
  - Converts digital data into signals (electrical, optical, or radio).
- **Examples**: Ethernet cables, Wi-Fi signals, USB.
- **Data Unit**: **Bits**.

### 2. Data Link Layer
- **Main Function**: Provides **node-to-node** data transfer and error detection.
- **Key Responsibilities**:
  - Framing, MAC addressing, error detection (CRC).
  - Manages access to the physical media (MAC sublayer).
- **Examples**: Ethernet (IEEE 802.3), Wi-Fi (IEEE 802.11).
- **Data Unit**: **Frames**.

### 3. Network Layer
- **Main Function**: Handles **routing** and forwarding of data packets.
- **Key Responsibilities**:
  - Logical addressing (IP addresses).
  - Path determination and packet forwarding.
- **Examples**: IP (Internet Protocol), ICMP, routers.
- **Data Unit**: **Packets**.

### 4. Transport Layer
- **Main Function**: Ensures **end-to-end** communication and data integrity.
- **Key Responsibilities**:
  - Flow control, error recovery, and segmentation.
  - Connection-oriented (TCP) or connectionless (UDP) communication.
- **Examples**: TCP, UDP.
- **Data Unit**: **Segments** (TCP) / **Datagrams** (UDP).

### 5. Session Layer
- **Main Function**: Manages **sessions** between applications.
- **Key Responsibilities**:
  - Establishes, maintains, and terminates connections.
  - Synchronization and checkpointing.
- **Examples**: NetBIOS, RPC, PPTP.
- **Data Unit**: **Data**.

### 6. Presentation Layer
- **Main Function**: Translates, encrypts, and compresses data.
- **Key Responsibilities**:
  - Data formatting, encryption (SSL/TLS), compression.
  - Ensures data is readable by the application layer.
- **Examples**: JPEG, MPEG, SSL.
- **Data Unit**: **Data**.

### 7. Application Layer
- **Main Function**: Provides **network services** to end-user applications.
- **Key Responsibilities**:
  - User interfaces, email, file transfer, and remote login.
- **Examples**: HTTP, FTP, SMTP, DNS.
- **Data Unit**: **Data**.

---

## Summary Table


OSI Model Layers


| Layer          | Function                          | Data Unit   | Examples               |
|----------------|-----------------------------------|-------------|------------------------|
| Physical       | Transmits raw bits                | Bits        | Ethernet, Wi-Fi        |
| Data Link      | Node-to-node transfer, framing    | Frames      | Ethernet, Wi-Fi        |
| Network        | Routing and forwarding            | Packets     | IP, ICMP, routers      |
| Transport      | End-to-end communication          | Segments    | TCP, UDP               |
| Session         | Manages sessions                  | Data        | NetBIOS, RPC           |
| Presentation   | Data translation/encryption       | Data        | JPEG, SSL              |
| Application    | Network services for applications | Data        | HTTP, FTP, SMTP        |

---

## Data Flow Explanation
- Data moves **down** the layers at the sender (encapsulation) and **up** at the receiver (decapsulation).
- Each layer adds its own **header** (and sometimes trailer) to the data.

# 2. TCP/IP Model

## Introduction
- The **TCP/IP Model** is a simplified, practical model used for internet communication.
- **Purpose**: To standardize how data is transmitted across networks, especially the internet.
- **Layers**: 4 layers (Application, Transport, Internet, Network Access).

---

## Layer-wise Explanation

### 1. Application Layer
- **Function**: Combines OSI’s **Application, Presentation, and Session** layers.
- **Responsibilities**:
  - Provides network services to applications.
  - Handles protocols for specific applications.
- **Examples**: HTTP, FTP, SMTP, DNS.

### 2. Transport Layer
- **Function**: Ensures **end-to-end** communication.
- **Responsibilities**:
  - Flow control, error recovery, and segmentation.
  - Connection-oriented (TCP) or connectionless (UDP).
- **Examples**: TCP, UDP.

### 3. Internet Layer
- **Function**: Handles **routing** and addressing.
- **Responsibilities**:
  - Logical addressing (IP), packet forwarding.
- **Examples**: IP, ICMP, routers.

### 4. Network Access Layer
- **Function**: Combines OSI’s **Physical and Data Link** layers.
- **Responsibilities**:
  - Transmits data over physical media.
  - MAC addressing, framing.
- **Examples**: Ethernet, Wi-Fi, PPP.

---

```
[Application Layer]
       ↓
[Transport Layer]
       ↓
[Internet Layer]
       ↓
[Network Access Layer]
```

# 1. Common Network Protocols

## Introduction
- **Network protocols** are rules and conventions that enable communication between devices on a network.
- They define how data is formatted, transmitted, and received.

---

## Key Concepts

### HTTP / HTTPS
- **Full Form**: HyperText Transfer Protocol / HyperText Transfer Protocol Secure
- **Purpose**: Transfer web pages and data over the internet.
- **Layer**: Application (OSI/TCP-IP)
- **Key Features**:
  - **HTTP**: Unencrypted, uses port 80.
  - **HTTPS**: Encrypted (SSL/TLS), uses port 443.
- **Example**: Loading a website (e.g., `https://www.example.com`).

### FTP
- **Full Form**: File Transfer Protocol
- **Purpose**: Transfer files between a client and a server.
- **Layer**: Application (OSI/TCP-IP)
- **Key Features**:
  - Uses two connections: control (port 21) and data (port 20).
  - Supports authentication.
- **Example**: Uploading files to a web server.

### SMTP
- **Full Form**: Simple Mail Transfer Protocol
- **Purpose**: Send and receive emails.
- **Layer**: Application (OSI/TCP-IP)
- **Key Features**:
  - Uses port 25.
  - Works with POP3/IMAP for email retrieval.
- **Example**: Sending an email via Gmail.

### DNS
- **Full Form**: Domain Name System
- **Purpose**: Translate domain names to IP addresses.
- **Layer**: Application (OSI/TCP-IP)
- **Key Features**:
  - Uses port 53.
  - Hierarchical and distributed database.
- **Example**: Converting `google.com` to `172.217.160.46`.

### DHCP
- **Full Form**: Dynamic Host Configuration Protocol
- **Purpose**: Automatically assign IP addresses to devices.
- **Layer**: Application (OSI/TCP-IP)
- **Key Features**:
  - Uses ports 67 and 68.
  - Leases IP addresses for a limited time.
- **Example**: Connecting a laptop to a Wi-Fi network.

### TCP
- **Full Form**: Transmission Control Protocol
- **Purpose**: Provide reliable, ordered, and error-checked data delivery.
- **Layer**: Transport (OSI/TCP-IP)
- **Key Features**:
  - Connection-oriented.
  - Uses acknowledgments and retransmissions.
- **Example**: Downloading a file from the internet.

### UDP
- **Full Form**: User Datagram Protocol
- **Purpose**: Provide fast, connectionless data transmission.
- **Layer**: Transport (OSI/TCP-IP)
- **Key Features**:
  - No acknowledgments or retransmissions.
  - Low overhead, faster than TCP.
- **Example**: Video streaming or online gaming.

# 2. IEEE 802.3 Standard

## Introduction
- **IEEE**: Institute of Electrical and Electronics Engineers.
- **IEEE 802.3**: Standard for **Ethernet** networks.

---

## Key Concepts
- **Purpose**: Define physical and data link layer specifications for wired Ethernet.
- **Importance**: Ensures compatibility and interoperability between Ethernet devices.
- **Relation with Ethernet**: IEEE 802.3 is the technical standard that defines Ethernet.

---

# 3. Ethernet

## Introduction
- **Definition**: A wired network technology used to connect devices in a LAN.

---

## Key Concepts
- **Basic Working**:
  - Devices send data in **frames**.
  - Uses **CSMA/CD** (Carrier Sense Multiple Access with Collision Detection) to manage data transmission.
- **Frames**: Data packets containing source/destination **MAC addresses** and payload.
- **MAC Address**: Unique identifier for network interfaces (e.g., `00:1A:2B:3C:4D:5E`).
- **CSMA/CD**: Devices listen for a quiet network before transmitting to avoid collisions.

---

## Real-World Analogy
- Think of Ethernet as a **postal system**:
  - **Frames** are letters.
  - **MAC addresses** are postal addresses.
  - **CSMA/CD** is like checking if the postman is free before handing over a letter.

---

# 4. Fast Ethernet

## Introduction
- **Definition**: An evolution of Ethernet with a speed of **100 Mbps**.

---

## Key Concepts
- **Speed**: 100 Mbps (10x faster than traditional Ethernet).
- **Improvements**: Higher bandwidth, backward compatible with 10BASE-T.
- **Standards**: **100BASE-TX** (uses Cat 5 cables), **100BASE-FX** (fiber optic).

---

# 5. Gigabit Ethernet

## Introduction
- **Definition**: Further evolution of Ethernet with a speed of **1 Gbps**.

---

## Key Concepts
- **Speed**: 1 Gbps (10x faster than Fast Ethernet).
- **Key Features**: Higher throughput, supports full-duplex communication.
- **Standards**: **1000BASE-T** (Cat 5e/6 cables), **1000BASE-SX/LX** (fiber optic).
